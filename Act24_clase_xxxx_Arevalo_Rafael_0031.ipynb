```python
# Programa de libros (2 objetos)
# Autor: Arevalo Rafael

class Libro:
    def __init__(self, titulo, autor, genero, paginas):
        self.titulo = titulo
        self.autor = autor
        self.genero = genero
        self.paginas = paginas

    def abrir(self):
        print("Abriendo libro:", self.titulo)

    def marcarPagina(self, pagina):
        if pagina >= 1 and pagina <= self.paginas:
            print("Página", pagina, "marcada en", self.titulo)
        else:
            print("Página fuera de rango")

    def buscarCapitulo(self, capitulo):
        print("Buscando capítulo", capitulo, "en", self.titulo)


# -------- LIBRO 1 --------
print("LIBRO 1")
titulo1 = input("Título: ")
autor1 = input("Autor: ")
genero1 = input("Género: ")
paginas1 = int(input("Número de páginas: "))

libro1 = Libro(titulo1, autor1, genero1, paginas1)


# -------- LIBRO 2 --------
print("\nLIBRO 2")
titulo2 = input("Título: ")
autor2 = input("Autor: ")
genero2 = input("Género: ")
paginas2 = int(input("Número de páginas: "))

libro2 = Libro(titulo2, autor2, genero2, paginas2)


# -------- MOSTRAR DATOS --------
print("\nDATOS LIBRO 1")
print("Título:", libro1.titulo)
print("Autor:", libro1.autor)

print("\nDATOS LIBRO 2")
print("Título:", libro2.titulo)
print("Autor:", libro2.autor)


# -------- ACCIONES --------
print("\nACCIONES LIBRO 1")
libro1.abrir()
p = int(input("Página a marcar: "))
libro1.marcarPagina(p)
c = input("Capítulo a buscar: ")
libro1.buscarCapitulo(c)

print("\nACCIONES LIBRO 2")
libro2.abrir()
p = int(input("Página a marcar: "))
libro2.marcarPagina(p)
c = input("Capítulo a buscar: ")
libro2.buscarCapitulo(c)


print("\nPrograma terminado y generado por Arevalo Rafael 0031")
```
